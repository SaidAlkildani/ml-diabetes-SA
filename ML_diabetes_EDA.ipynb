```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the clean dataset
diabetes_df = pd.read_csv('ML_diabetes_clean.csv')
```

```python
diabetes_df.info()
```

```python
# Data correlation with target variable
correlation = diabetes_df.corr()['outcome'].sort_values(ascending=False)
print("Correlation with target variable:\n", correlation)
```

```python
# pair plotting related features
## Removing zeros to visulise the correlations better
columns_to_fix = ['bmi', 'skinthickness', 'insulin', 'glucose']
diabetes_df_clean = diabetes_df[~(diabetes_df[columns_to_fix] == 0).any(axis=1)]

## BMI and skin thickness, strong correlation
sns.pairplot(diabetes_df_clean, vars= ["bmi","skinthickness"], hue='outcome', diag_kind='kde')
## Glucose and insulin levels, strong correlation
sns.pairplot(diabetes_df_clean, vars= ["glucose","insulin"], hue='outcome', diag_kind='kde')
```

## Insights
The features seems to have high correlations with diabetes risk. 

Highly correlated values should be considered when fitting linear/logistic models. This could lead to overfitting. How better do desicion tree models behave?
