# Fetching Data

```python
# Importing libraries
import pandas as pd
from sqlalchemy import create_engine
import os
from dotenv import load_dotenv
```

```python
# Loading .env file
load_dotenv()

DATABASE = os.getenv('DATABASE')
USER_DB = os.getenv('USER_DB')
PASSWORD = os.getenv('PASSWORD')
HOST = os.getenv('HOST')
PORT = os.getenv('PORT')
```

```python
# Read the database string from the .env
load_dotenv()

DB_STRING = os.getenv('DB_STRING')

db = create_engine(DB_STRING)
```

```python
# Import the data to a pandas dataframe
query_string = "set schema 'diabetes'; SELECT * FROM patient p LEFT JOIN blood_metrics bd ON p.id = bd.patientid LEFT JOIN skin s ON p.id = s.patientid LEFT JOIN pedigree_outcome po ON p.id = po.PatientId where bd.measurement_date = '2022-12-13';"
df_diabetes = pd.read_sql(query_string, db)
```

```python
# Test if the data is fetched correctly
df_diabetes.shape
df_diabetes.head()
```

```python
# Export the data to a csv-file
df_diabetes.to_csv('ML_diabetes.csv',index=False)
```
